```python
from clasificador_paquetes.contracts import Entrada, Salida
import pandas as pd

csv = pd.read_csv("data/raw/paquetes.csv")
l = []
for i, row in csv.iterrows():
    cleaned_row = Entrada.model_validate(row.to_dict())
    l.append(cleaned_row)
```

```python
import math
print(cleaned_row.peso_kg)
a = round(cleaned_row.peso_kg, 1)
print(a)
print(cleaned_row.distancia_km)
b = math.floor(cleaned_row.distancia_km)
print(b)
```

```python
import joblib

model = joblib.load("models/modelo.joblib")
v = model.predict([[a,b]])
p = model.predict_proba([[a,b]])
print(v)
print(p[0][1])
value = None
if p[0][0] < p[0][1]:
    value = "urgente"
else:
    value = "normal"
```
